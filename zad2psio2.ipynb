```python
#zad 2

import numpy as np
import matplotlib.pyplot as plt

def sinus(A, f, fi, fs, N):
    """
    A - amplituda
    f - częstotliwość (Hz)
    fi - faza (w radianach)
    fs - próbkowanie (Hz)
    N - liczba próbek
    """
    n = np.arange(N)
    y = A * np.sin(2 * np.pi * f * n / fs + fi)
    return n, y

fs = 100
N = 50

#zmiana amplitudy
n1,y1=sinus(1,5,0,fs,N)
n2,y2=sinus(2,5,0,fs,N)

#zmiana czestotliwosci
n3,y3=sinus(1,5,0,fs,N)
n4,y4=sinus(1,10,0,fs,N)

#zmiana fazy
n5,y5=sinus(1,5,0,fs,N)
n6,y6=sinus(1,5,np.pi/2,fs,N)


plt.figure(figsize=(10,6))
plt.subplot(3,1,1)
plt.stem(n1,y1,linefmt='b-', markerfmt='bo', basefmt='k-',label='A=1')
plt.stem(n2,y2,linefmt='r-', markerfmt='ro', basefmt='k-',label='A=2')
plt.title("Wpływ amplitudy")
plt.legend()
plt.grid()

plt.subplot(3,1,2)
plt.stem(n3, y3, linefmt='b-', markerfmt='bo', basefmt='k-', label='f=5')
plt.stem(n4, y4, linefmt='r-', markerfmt='ro', basefmt='k-', label='f=10')
plt.title("Wpływ częstotliwości")
plt.legend()
plt.grid()

plt.subplot(3,1,3)
plt.stem(n5,y5,linefmt='b-', markerfmt='bo', basefmt='k-',label='fi=0')
plt.stem(n6,y6,linefmt='r-', markerfmt='ro', basefmt='k-',label='fi=pi/2')
plt.title("Wpływ fazy")
plt.legend()
plt.grid()

# Zapis danych wykresow z komorki 1 do pozniejszego laczenia
task2_cell1_data = {
    "fs": fs,
    "N": N,
    "n1": n1.copy(), "y1": y1.copy(),
    "n2": n2.copy(), "y2": y2.copy(),
    "n3": n3.copy(), "y3": y3.copy(),
    "n4": n4.copy(), "y4": y4.copy(),
    "n5": n5.copy(), "y5": y5.copy(),
    "n6": n6.copy(), "y6": y6.copy(),
}
```

```python
# zad 2

import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio,display

def sinus(A, f, fi, fs, N):
    """
    A - amplituda
    f - częstotliwość (Hz)
    fi - faza (w radianach)
    fs - próbkowanie (Hz)
    N - liczba próbek
    """
    n = np.arange(N)
    y = A * np.sin(2 * np.pi * f * n / fs + fi)
    return n, y

fs = 44100
duration = 2
N = int(fs * duration)

# zmiana amplitudy
n1,y1 = sinus(0.3, 440, 0, fs, N)
n2,y2 = sinus(0.8, 440, 0, fs, N)

# zmiana częstotliwości
n3,y3 = sinus(0.5, 220, 0, fs, N)
n4,y4 = sinus(0.5, 880, 0, fs, N)

# zmiana fazy
n5,y5 = sinus(0.5, 440, 0, fs, N)
n6,y6 = sinus(0.5, 440, np.pi/2, fs, N)

plt.figure(figsize=(10,6))

plt.subplot(3,1,1)
plt.plot(n1[:1000], y1[:1000], label='A=0.3')
plt.plot(n2[:1000], y2[:1000], label='A=0.8')
plt.title("Wpływ amplitudy")
plt.legend()
plt.grid()

plt.subplot(3,1,2)
plt.plot(n3[:1000], y3[:1000], label='f=220')
plt.plot(n4[:1000], y4[:1000], label='f=880')
plt.title("Wpływ częstotliwości")
plt.legend()
plt.grid()

plt.subplot(3,1,3)
plt.plot(n5[:1000], y5[:1000], label='fi=0')
plt.plot(n6[:1000], y6[:1000], label='fi=pi/2')
plt.title("Wpływ fazy")
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()


#odsłuchy

print("Amplituda:")
display(Audio(y1, rate=fs))
display(Audio(y2, rate=fs))

print("Częstotliwość:")
display(Audio(y3, rate=fs))
display(Audio(y4, rate=fs))

print("Faza:")
display(Audio(y5, rate=fs))
display(Audio(y6, rate=fs))

# Zapis danych wykresow z komorki 2 do pozniejszego laczenia
task2_cell2_data = {
    "fs": fs,
    "N": N,
    "n1": n1.copy(), "y1": y1.copy(),
    "n2": n2.copy(), "y2": y2.copy(),
    "n3": n3.copy(), "y3": y3.copy(),
    "n4": n4.copy(), "y4": y4.copy(),
    "n5": n5.copy(), "y5": y5.copy(),
    "n6": n6.copy(), "y6": y6.copy(),
}
```

```python
# Zbiorcze zestawienie wykresow z komorki 1 i komorki 2 w jednym oknie
if "task2_cell1_data" not in globals() or "task2_cell2_data" not in globals():
    raise RuntimeError(
        "Brak danych do laczenia wykresow. Uruchom najpierw komorke 1 i komorke 2, "
        "a potem uruchom te komorke ponownie."
    )

c1 = task2_cell1_data
c2 = task2_cell2_data

fig, axs = plt.subplots(3, 2, figsize=(15, 10), constrained_layout=True)
fig.suptitle("Zadanie 2 - zestawienie wykresow z komorki 1 i 2", fontsize=14)

# Wplyw amplitudy
axs[0, 0].stem(c1["n1"], c1["y1"], linefmt="b-", markerfmt="bo", basefmt="k-", label="A=1")
axs[0, 0].stem(c1["n2"], c1["y2"], linefmt="r-", markerfmt="ro", basefmt="k-", label="A=2")
axs[0, 0].set_title("Wplyw amplitudy (komorka 1)")
axs[0, 0].set_xlabel("Numer probki n [-]")
axs[0, 0].set_ylabel("Amplituda [1]")
axs[0, 0].grid(True, alpha=0.3)
axs[0, 0].legend()

axs[0, 1].plot(c2["n1"][:1000] / c2["fs"], c2["y1"][:1000], label="A=0.3", linewidth=1.5)
axs[0, 1].plot(c2["n2"][:1000] / c2["fs"], c2["y2"][:1000], label="A=0.8", linewidth=1.5)
axs[0, 1].set_title("Wplyw amplitudy (komorka 2)")
axs[0, 1].set_xlabel("Czas t [s]")
axs[0, 1].set_ylabel("Amplituda [1]")
axs[0, 1].grid(True, alpha=0.3)
axs[0, 1].legend()

# Wplyw czestotliwosci
axs[1, 0].stem(c1["n3"], c1["y3"], linefmt="b-", markerfmt="bo", basefmt="k-", label="f=5")
axs[1, 0].stem(c1["n4"], c1["y4"], linefmt="r-", markerfmt="ro", basefmt="k-", label="f=10")
axs[1, 0].set_title("Wplyw czestotliwosci (komorka 1)")
axs[1, 0].set_xlabel("Numer probki n [-]")
axs[1, 0].set_ylabel("Amplituda [1]")
axs[1, 0].grid(True, alpha=0.3)
axs[1, 0].legend()

axs[1, 1].plot(c2["n3"][:1000] / c2["fs"], c2["y3"][:1000], label="f=220 Hz", linewidth=1.5)
axs[1, 1].plot(c2["n4"][:1000] / c2["fs"], c2["y4"][:1000], label="f=880 Hz", linewidth=1.5)
axs[1, 1].set_title("Wplyw czestotliwosci (komorka 2)")
axs[1, 1].set_xlabel("Czas t [s]")
axs[1, 1].set_ylabel("Amplituda [1]")
axs[1, 1].grid(True, alpha=0.3)
axs[1, 1].legend()

# Wplyw fazy
axs[2, 0].stem(c1["n5"], c1["y5"], linefmt="b-", markerfmt="bo", basefmt="k-", label="fi=0")
axs[2, 0].stem(c1["n6"], c1["y6"], linefmt="r-", markerfmt="ro", basefmt="k-", label="fi=pi/2")
axs[2, 0].set_title("Wplyw fazy (komorka 1)")
axs[2, 0].set_xlabel("Numer probki n [-]")
axs[2, 0].set_ylabel("Amplituda [1]")
axs[2, 0].grid(True, alpha=0.3)
axs[2, 0].legend()

axs[2, 1].plot(c2["n5"][:1000] / c2["fs"], c2["y5"][:1000], label="fi=0", linewidth=1.5)
axs[2, 1].plot(c2["n6"][:1000] / c2["fs"], c2["y6"][:1000], label="fi=pi/2", linewidth=1.5)
axs[2, 1].set_title("Wplyw fazy (komorka 2)")
axs[2, 1].set_xlabel("Czas t [s]")
axs[2, 1].set_ylabel("Amplituda [1]")
axs[2, 1].grid(True, alpha=0.3)
axs[2, 1].legend()

plt.show()
```
