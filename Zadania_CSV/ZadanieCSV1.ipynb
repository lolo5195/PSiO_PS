```python
import cv2                       # biblioteka OpenCV
import matplotlib.pyplot as plt  # rysowanie wykresow

sciezka = "../PILLOW/chlopak.JPG"                    # sciezka do obrazu na dysku
obraz = cv2.imread(sciezka)                          # wczytanie obrazu (format kolorow BGR)
obraz_rgb = cv2.cvtColor(obraz, cv2.COLOR_BGR2RGB)   # zamiana BGR -> RGB do poprawnego wyswietlenia

kolory = ("b", "g", "r")                             # kanaly koloru w kolejnosci OpenCV

plt.figure(figsize=(12, 5))                          # wspolne okno na obraz i histogram

plt.subplot(1, 2, 1)                                 # po lewej: wczytany obraz
plt.imshow(obraz_rgb)
plt.title("Wczytany obraz: chlopak.JPG")
plt.axis("off")

plt.subplot(1, 2, 2)                                 # po prawej: histogram
for i, kolor in enumerate(kolory):                   # petla po 3 kanalach koloru
    histogram = cv2.calcHist([obraz], [i], None, [256], [0, 256])  # zliczenie pikseli kanalu
    plt.plot(histogram, color=kolor)
plt.title("Histogram obrazu")
plt.xlabel("Jasnosc piksela (0-255)")
plt.ylabel("Liczba pikseli")
plt.xlim([0, 256])

plt.tight_layout()                                   # estetyczne rozmieszczenie
plt.show()                                           # wyswietlenie wynikow
```
