```python
import cv2                       # biblioteka OpenCV
import matplotlib.pyplot as plt  # rysowanie histogramów

sciezka = "../PILLOW/chlopak.JPG"                  # ścieżka do obrazu
obraz = cv2.imread(sciezka, cv2.IMREAD_GRAYSCALE)  # obraz bazowy w skali szarości
obraz_eq = cv2.equalizeHist(obraz)                 # obraz po wyrównaniu histogramu

plt.figure(figsize=(12, 5))  # wspólne okno na oba histogramy

plt.subplot(1, 2, 1)                                    # histogram bazowy (po lewej)
plt.hist(obraz.ravel(), 256, [0, 256], color="gray")
plt.title("Histogram bazowy")
plt.xlabel("Jasność piksela (0-255)")
plt.ylabel("Liczba pikseli")

plt.subplot(1, 2, 2)                                    # histogram po korekcie (po prawej)
plt.hist(obraz_eq.ravel(), 256, [0, 256], color="gray")
plt.title("Histogram po wyrównaniu")
plt.xlabel("Jasność piksela (0-255)")
plt.ylabel("Liczba pikseli")

plt.tight_layout()  # czytelne rozmieszczenie wykresów
plt.show()
```
